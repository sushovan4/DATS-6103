Then fix `total_price` so that an item with no `"qty"` counts as a quantity of
one. Do not edit `cart`.

````python
def total_price(items):
    total = 0
    for item in items:
        total += item["price"] * item["qty"]
    return total

cart = [{"price": 2.5, "qty": 4}, {"price": 3.0}, {"price": 1.25, "qty": 2}]
````
